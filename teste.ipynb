# Resumo Completo — Séries Temporais

## 1. Introdução

Uma **série temporal** é uma sequência de observações coletadas ao longo do tempo em intervalos regulares.

Exemplos:

| Série              | Frequência |
| ------------------ | ---------- |
| Vendas de uma loja | Diária     |
| Cotação de ações   | Minutária  |
| Temperatura        | Horária    |
| Número de clientes | Mensal     |

O principal objetivo da análise de séries temporais é:

* compreender padrões históricos;
* identificar tendências;
* detectar sazonalidades;
* realizar previsões futuras.

---

# 2. Conceitos Básicos

## Tendência (Trend)

Representa o comportamento de longo prazo da série.

### Crescente

| Tempo | Valor |
| ----- | ----- |
| 1     | 10    |
| 2     | 15    |
| 3     | 20    |
| 4     | 25    |

### Decrescente

| Tempo | Valor |
| ----- | ----- |
| 1     | 25    |
| 2     | 20    |
| 3     | 15    |
| 4     | 10    |

---

## Sazonalidade

Padrão que se repete periodicamente.

Exemplos:

| Série      | Sazonalidade |
| ---------- | ------------ |
| Sorvetes   | Verão        |
| Energia    | Horário      |
| E-commerce | Black Friday |

Formalmente:

[
y_t = S_t
]

onde:

* (S_t) = componente sazonal.

---

## Ruído

Parte aleatória da série.

[
\varepsilon_t
]

Idealmente:

[
\varepsilon_t \sim N(0,\sigma^2)
]

---

## Ciclo

Oscilações de longo prazo.

Exemplo:

* economia
* inflação
* desemprego

---

## Granularidade

Representa o nível temporal dos dados.

| Granularidade | Exemplo     |
| ------------- | ----------- |
| Minuto        | sensores    |
| Hora          | temperatura |
| Dia           | vendas      |
| Semana        | faturamento |
| Mês           | receita     |

---

## Estacionariedade

Conceito mais importante da disciplina.

Uma série estacionária possui:

### Média constante

[
E[y_t]=\mu
]

### Variância constante

[
Var(y_t)=\sigma^2
]

### Covariância constante

[
Cov(y_t,y_{t-k})
]

depende apenas do lag.

---

## STL

Seasonal Trend using Loess

Decompõe a série em:

[
y_t=T_t+S_t+R_t
]

Onde:

* (T_t) = tendência
* (S_t) = sazonalidade
* (R_t) = ruído

---

# 3. Estatísticas Dinâmicas

## Média Móvel (SMA)

Suaviza a série.

[
SMA_t=\frac{1}{k}\sum_{i=0}^{k-1} y_{t-i}
]

### Exemplo

Janela = 3

| Série |
| ----- |
| 10    |
| 12    |
| 15    |

[
SMA=\frac{10+12+15}{3}
]

[
SMA=12.33
]

---

## Média Móvel Exponencial (EMA)

Dá mais peso aos valores recentes.

[
EMA_t=\alpha y_t + (1-\alpha) EMA_{t-1}
]

onde:

[
0<\alpha<1
]

### Exemplo

[
\alpha=0.8
]

[
EMA=0.8(12)+0.2(10)
]

[
EMA=11.6
]

---

# 4. Métricas de Erro

## MAE

Erro absoluto médio.

[
MAE=
\frac1n
\sum |y_t-\hat y_t|
]

### Vantagens

* simples
* interpretável

### Desvantagens

* não penaliza erros grandes

---

## MSE

[
MSE=
\frac1n
\sum(y_t-\hat y_t)^2
]

### Vantagens

* penaliza erros grandes

### Desvantagens

* sensível a outliers

---

## RMSE

[
RMSE=\sqrt{MSE}
]

Mesmo significado do MAE porém penalizando mais erros grandes.

---

## MAPE

[
MAPE=
\frac{100}{n}
\sum
\left|
\frac{y_t-\hat y_t}{y_t}
\right|
]

### Problema

Não funciona bem próximo de zero.

---

# 5. Base Models

Segundo a disciplina, os base models são utilizados como **baseline** para verificar se modelos mais complexos realmente agregam valor.

Na pipeline:

```text
Série
 ↓
Base Models
 ↓
Avaliação
 ↓
SARIMA
 ↓
Comparação
```

Na atividade N2 é obrigatório comparar SARIMA com todos os base models. 

---

# 5.1 Média Histórica

Prevê sempre a média da série.

[
\hat y_t=
\bar y
]

[
\bar y=
\frac1n
\sum y_i
]

### Melhor cenário

Séries estáveis.

### Pontos fortes

* extremamente simples

### Pontos fracos

* ignora tendência
* ignora sazonalidade

---

# 5.2 Média Acumulada

Usa todos os pontos anteriores.

[
\hat y_t=
\frac1{t-1}
\sum_{i=1}^{t-1}y_i
]

### Melhor cenário

Séries sem mudanças bruscas.

### Vantagens

* adapta-se ao longo do tempo

### Desvantagens

* reage lentamente

---

# 5.3 Média Móvel (SMA)

[
\hat y_t=
\frac{
y_{t-1}+...+y_{t-k}
}{k}
]

### Melhor cenário

Ruído elevado.

### Vantagens

* suaviza oscilações

### Desvantagens

* atraso na tendência

---

# 5.4 Média Móvel Exponencial (EMA)

[
EMA_t=
\alpha y_t+
(1-\alpha)EMA_{t-1}
]

### Melhor cenário

Mudanças recentes importantes.

### Vantagens

* responde rápido

### Desvantagens

* escolha do α

---

# 5.5 Taxa de Variação

Calcula crescimento percentual.

[
taxa=
\frac{
y_t-y_{t-k}
}{
y_{t-k}
}
]

Previsão:

[
\hat y_{t+1}
============

y_t(1+taxa)
]

### Melhor cenário

Crescimento percentual consistente.

### Vantagens

* captura crescimento

### Desvantagens

* instável em séries voláteis

---

# 5.6 Seasonal Naive

Copia o valor observado no mesmo período sazonal.

[
\hat y_t=y_{t-s}
]

onde:

* (s) = período sazonal.

### Exemplo

Dados mensais:

[
\hat y_{jan2026}
================

y_{jan2025}
]

### Melhor cenário

Sazonalidade forte.

### Vantagens

* excelente baseline

### Desvantagens

* ignora tendência

---

# 5.7 Delta (Drift)

Calcula tendência média.

[
\delta=
\frac{
y_t-y_{t-k}
}{
k
}
]

Previsão:

[
\hat y_{t+1}
============

y_t+\delta
]

### Melhor cenário

Tendência linear.

### Vantagens

* simples
* interpreta tendência

### Desvantagens

* falha em sazonalidade

---

# 6. Diferenciação

Necessária porque:

* AR
* MA
* ARMA

exigem estacionariedade. 

---

## Primeira Diferença

[
y'_t
====

y_t-y_{t-1}
]

y'*t=y_t-y*{t-1}

Remove tendência linear.

---

## Segunda Diferença

[
y''*t=
(y_t-y*{t-1})
-------------

(y_{t-1}-y_{t-2})
]

Remove tendência quadrática.

---

## Diferença de Ordem d

[
(1-B)^d y_t
]

onde:

* (B) = operador de atraso.

---

## Diferença Sazonal

[
y'_t
====

y_t-y_{t-s}
]

y'*t=y_t-y*{t-s}

Exemplo:

[
s=12
]

para dados mensais.

---

# 7. Autocorrelação

## ACF

Função de autocorrelação.

Mede relação entre:

[
y_t
]

e

[
y_{t-k}
]

Ajuda na escolha de:

[
q
]

e

[
Q
]

---

## PACF

Autocorrelação parcial.

Remove efeito intermediário dos lags.

Ajuda na escolha de:

[
p
]

e

[
P
]

# 8. Testes de Estacionariedade

## ADF

Hipótese nula:

[
H_0:
\text{série não estacionária}
]

Se:

[
p<0.05
]

rejeitamos (H_0).

---

## KPSS

Hipótese nula:

[
H_0:
\text{série estacionária}
]

Se:

[
p<0.05
]

rejeitamos (H_0).

---

## Interpretação conjunta

| ADF            | KPSS           | Resultado        |
| -------------- | -------------- | ---------------- |
| Rejeita H0     | Não rejeita H0 | Estacionária     |
| Não rejeita H0 | Rejeita H0     | Não estacionária |

# 9. AutoReg (AR)

Modelo autoregressivo.

Prevê usando valores passados.

[
y_t=
c+
\phi_1y_{t-1}
+\phi_2y_{t-2}
+\cdots
+\phi_py_{t-p}
+\varepsilon_t
]

y_t=c+\phi_1y_{t-1}+\phi_2y_{t-2}+\cdots+\phi_py_{t-p}+\varepsilon_t

Parâmetro:

* p = número de lags

---

# 10. ARIMA

Combina:

* AR
* Integração (d)
* MA

[
ARIMA(p,d,q)
]

onde:

| Parâmetro | Significado   |
| --------- | ------------- |
| p         | AR            |
| d         | diferenciação |
| q         | MA            |

---

## Componente MA

[
y_t=
\mu+
\varepsilon_t+
\theta_1\varepsilon_{t-1}
+\cdots+
\theta_q\varepsilon_{t-q}
]

---

## Modelo Completo

[
ARIMA(p,d,q)
]

Fluxo:

```text
Série
 ↓
Diferenciação d
 ↓
AR(p)
 ↓
MA(q)
```

---

# 11. SARIMA

Modelo principal da disciplina.

[
SARIMA(p,d,q)(P,D,Q,m)
]

SARIMA(p,d,q)(P,D,Q,m)

---

## Componentes

### Parte Não Sazonal

| Parâmetro | Significado |
| --------- | ----------- |
| p         | AR          |
| d         | diferença   |
| q         | MA          |

---

### Parte Sazonal

| Parâmetro | Significado       |
| --------- | ----------------- |
| P         | AR sazonal        |
| D         | diferença sazonal |
| Q         | MA sazonal        |
| m         | período sazonal   |

---

## Exemplo

[
SARIMA(0,1,2)(0,1,2,7)
]

Modelo encontrado no exemplo da aula.
